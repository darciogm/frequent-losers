```python
import pandas as pd

df_won_1 = pd.read_csv('/Users/dvacchini/Library/CloudStorage/Dropbox/Projects_Brazilian_data/Dario_vacchini/data/raw/item_class/reg_won_t_minus_1_by_item_class_1.csv')
df_won_2 = pd.read_csv('/Users/dvacchini/Library/CloudStorage/Dropbox/Projects_Brazilian_data/Dario_vacchini/data/raw/item_class/reg_won_t_minus_1_by_item_class_2.csv')

df_last = pd.read_csv('/Users/dvacchini/Library/CloudStorage/Dropbox/Projects_Brazilian_data/Dario_vacchini/data/raw/item_class/reg_last_bid_by_item_class.csv')

df_90won = pd.read_csv('/Users/dvacchini/Library/CloudStorage/Dropbox/Projects_Brazilian_data/Dario_vacchini/data/raw/item_class/reg_share_won90_by_item_class_1.csv')

df_won = pd.concat([df_won_1, df_won_2], ignore_index=True).drop_duplicates()
df_won = df_won.loc[df_won['Variable'] != '1.flagvencedor#c.MV']
df_last = df_last.loc[df_last['Variable'] != '1.flagvencedor#c.MV']
df_90won = df_90won.loc[df_90won['Variable'] != '1.flagvencedor#c.MV']

# Rename columns if needed to match the example
df_won = df_won.rename(columns={' Market Item': 'market_id', ' Coefficient': 'Incumbency', ' Std. Error': 'std_error', ' Sig_1pct': 'sig_1pct', ' Sig_5pct': 'sig_5pct', ' Sig_10pct': 'sig_10pct'})
df_last = df_last.rename(columns={' Market Item': 'market_id', ' Coefficient': 'Last Bid', ' Std. Error': 'std_error', ' Sig_1pct': 'sig_1pct', ' Sig_5pct': 'sig_5pct', ' Sig_10pct': 'sig_10pct'})
df_90won = df_90won.rename(columns={' Market Item': 'market_id', ' Coefficient': '90-Day Share Won', ' Std. Error': 'std_error', ' Sig_1pct': 'sig_1pct', ' Sig_5pct': 'sig_5pct', ' Sig_10pct': 'sig_10pct'})
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

#create scatterplot with regression line and confidence interval lines
sns.regplot(
    x=df_won['Incumbency'],
    y=df_last['Last Bid'],
    ci=None,
    color='dimgrey',                 
    scatter_kws={'s': 20, 'alpha': 0.7},  # Reduce dot size and add transparency
    line_kws={'color': 'black'}, # Optional: set regression line color
)

plt.xlabel('Incumbency')
plt.ylabel('Last Bid')
plt.title('Binscatter: Last Bid vs. Incumbency')
sns.despine()  # Clean up plot borders
plt.tight_layout()

# Save the plot to a file (e.g., PNG)
plt.savefig('/Users/dvacchini/Library/CloudStorage/Dropbox/Projects_Brazilian_data/Dario_vacchini/output/graphs/binscatter_incumbency_last_bid_item_class.pdf', dpi=300, bbox_inches='tight')

# Optional: Show the plot (if you want to display it)
plt.show()
```
